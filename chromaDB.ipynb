```python
from chromadb.utils import embedding_functions
import numpy as np
client = chromadb.Client()

embedding_function = embedding_functions.DefaultEmbeddingFunction()

collection = client.create_collection(
    name="products",
    embedding_function=embedding_function)
```

```python
### Query:
collection.add(
    ids=['1','2','3'],
    documents=['apple iphone smartphone', "dell laptop computers",'samsung andriod smartphone',])
```

```python
results=collection.query(query_texts=['mobile phone'],n_results=2)
```
